```python
import cv2 as cv
import numpy as np
from matplotlib import pyplot as plt
```

```python
#load input image
image = cv.imread('smurf.jpg')

#use cvtColor untuk conver ke grayscale
gray_image = cv.cvtColor(image, cv.COLOR_BGR2GRAY)

#disini kita coba dari library matplotlib
plt.subplot(1,2,1)
plt.imshow(cv.cvtColor(image, cv.COLOR_BGR2RGB))
plt.axis('off')
plt.title('Original')

plt.subplot(1,2,2)
plt.imshow(cv.cvtColor(gray_image, cv.COLOR_BGR2RGB))
plt.axis('off')
plt.title('Grayscale')

plt.tight_layout()
plt.show()
```
