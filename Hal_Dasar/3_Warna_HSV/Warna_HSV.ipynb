```python
import cv2 as cv
import numpy as np
from matplotlib import pyplot as plt
```

```python
img = cv.imread('flower.jpg')

hsv_img = cv.cvtColor(img, cv.COLOR_BGR2HSV)

plt.figure(figsize=(12,8))

plt.subplot(1,4,1)
plt.imshow(hsv_img)
plt.axis('off')
plt.title('HSV')

plt.subplot(1,4,2)
plt.imshow(cv.cvtColor(hsv_img[:,:,0], cv.COLOR_BGR2RGB))
plt.axis('off')
plt.title('HUE')

plt.subplot(1,4,3)
plt.imshow(cv.cvtColor(hsv_img[:,:,1], cv.COLOR_BGR2RGB))
plt.axis('off')
plt.title('SATURATION')

plt.subplot(1,4,4)
plt.imshow(cv.cvtColor(hsv_img[:,:,2], cv.COLOR_BGR2RGB))
plt.axis('off')
plt.title('VALUE')

plt.tight_layout()
plt.show()
```

Color space is HSV

H: 0 - 180

S: 0 - 255

V: 0 - 255
