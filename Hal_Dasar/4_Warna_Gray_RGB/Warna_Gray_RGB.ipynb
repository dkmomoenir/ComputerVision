```python
import cv2 as cv
import numpy as np
from matplotlib import pyplot as plt
```

```python
img = cv.imread('anime.jpg')

#BGR value untuk 0,0 pixel
B,G,R = img[10,50]
print(B,G,R)
print(img.shape)
```

```python
#convert ke grayscale
gray_img = cv.cvtColor(img, cv.COLOR_BGR2GRAY)
print(gray_img.shape)
print(gray_img[10,50])
```

```python
#menggunakan OpenCV 'split' untuk membagi gamber menjadi beberapa color index
B,G,R = cv.split(img)
print(B.shape)
```

```python
plt.figure(figsize=(10,8))

plt.subplot(1, 3, 1)
plt.imshow(cv.cvtColor(R, cv.COLOR_BGR2RGB))
plt.title('Red')
plt.axis('off')

plt.subplot(1,3,2)
plt.imshow(cv.cvtColor(G, cv.COLOR_BGR2RGB))
plt.title('Green')
plt.axis('off')

plt.subplot(1,3,3)
plt.imshow(cv.cvtColor(B, cv.COLOR_BGR2RGB))
plt.title('Blue')
plt.axis('off')

plt.tight_layout()
plt.show()
```

Merge with specific RGB

```python
B,G,R = cv.split(img)
zeros = np.zeros(img.shape[:2],dtype="uint8")
```

```python
plt.figure(figsize=(15,15))

plt.subplot(1,3,1)
plt.imshow(cv.merge([R,zeros,zeros]))
plt.axis('off')
plt.title('RED')

plt.subplot(1,3,2)
plt.imshow(cv.merge([zeros,G,zeros]))
plt.axis('off')
plt.title('GREEN')

plt.subplot(1,3,3)
plt.imshow(cv.merge([zeros,zeros,B]))
plt.axis('off')
plt.title('BLUE')

plt.tight_layout()
plt.show()
```
