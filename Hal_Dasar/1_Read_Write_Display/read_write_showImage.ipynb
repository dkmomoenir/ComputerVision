```python
import cv2 as cv
import numpy as np
from matplotlib import pyplot as plt
```

```python
#imread untuk upload gambar
img = cv.imread('flower.jpg')

#display image dengan matplotlib
#cvtColor untuk convert BGR ke RGB
plt.imshow(cv.cvtColor(img, cv.COLOR_BGR2RGB))
plt.title('Flower')

plt.show
```

```python
#untuk display variabel gambar dengan cv, dengan 'imshow'
#parameter1 untuk nama gambar, parameter 2 untuk variabel gambarnya

cv.imshow('Flower', img)
cv.waitKey(0)
```

```python
#untuk show image shape
print(img.shape)
```

```python
#karena image terlalu besar maka kita akan resize gambar tersebut ke 10%
scale = 0.10
width = int(img.shape[1]*scale)
height = int(img.shape[0]* scale)
dim = (width, height)

resized = cv.resize(img,dim,interpolation=cv.INTER_AREA)

cv.imshow('Flower',resized)
cv.waitKey(0)
cv.destroyAllWindows()
```

```python
#print setiap dimensi gambar

print(f'Tinggi dari gambar: {int(img.shape[0])} pixels')
print(f'Lebar dari gambar: {int(img.shape[1])} pixels')
```

```python
#Gunakan 'imwrite' untuk menulis ulang nama file dan gambar akan tersimpan
cv.imwrite('output.jpg',img)
```
