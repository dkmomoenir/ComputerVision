```python
import cv2 as cv
import numpy as np
from matplotlib import pyplot as plt
```

```python
#untuk membuat black image
image = np.zeros((512, 512, 3), np.uint8)
image_bw = np.zeros((512,512), np.uint8)
```

```python
plt.imshow(image)
plt.imshow(cv.cvtColor(image, cv.COLOR_BGR2RGB))
plt.axis('off')

plt.tight_layout()
plt.show()
```

Menggambar Garis

cv.line(image, koordinat awal, koordinat akhir, warna, ketebalan)

```python
img1 = np.zeros((512,512,3), np.uint8)

cv.line(img1, (0,0), (int(img1.shape[0]/2), int(img1.shape[0]/2)), (255,255,0),10)

cv.line(img1, (int(img1.shape[0]), 0), (int(img1.shape[0]/2), int(img1.shape[0]/2)), (0, 255, 255), 10)

cv.line(img1, (int(img1.shape[0]/2), int(img1.shape[0]/2)), (0, int(img1.shape[0])), (255, 0, 255), 10)

cv.line(img1, (int(img1.shape[0]/2), int(img1.shape[0]/2)), (int(img1.shape[0]), int(img1.shape[0])), (225, 225, 225), 10)

plt.imshow(cv.cvtColor(img1, cv.COLOR_BGR2RGB))
plt.show()
```

```python
img2 = np.zeros((512, 512, 3), np.uint8)

cv.rectangle(img2, (int(img2.shape[0]/2)-150, int(img2.shape[0]/2)-150),
             (int(img2.shape[0]/2)+150, int(img2.shape[0]/2)+150), (255, 200, 100), 2)

cv.line(img2, (int(img2.shape[0]/2)-150, int(img2.shape[0]/2)-150),
             (int(img2.shape[0]/2)+150, int(img2.shape[0]/2)+150), (0, 200, 100), 2)

plt.imshow(cv.cvtColor(img2, cv.COLOR_BGR2RGB))

plt.show
```

Cara menggambar Ellipse

Ellipse = cv.ellipse(image, kordinat tengah, axesLength, angle, starAngle, endANgle, warna, ketebalan)

```python
img4 = np.zeros((512, 512, 3), np.uint8)

e = cv.ellipse(img4, (int(img4.shape[0]/2), int(img4.shape[0]/2)), (200, 100), 45, 0, 360, (255, 150, 100), -1)
e = cv.ellipse(img4, (int(img4.shape[0]/2), int(img4.shape[0]/2)), (200 ,100), 315, 0, 360, (100, 150, 255), -1)

plt.imshow(cv.cvtColor(e, cv.COLOR_BGR2RGB))
plt.show()
```

Menggambar Polyline

polyline = cv.polyline(image,[pts], isClosed, warna, ketebalan)

```python
img4 = np.zeros((512, 512, 3), np.uint8)

pts = np.array([[100, 400], [250, 50], [400,400]], np.int32)
pts = pts.reshape((-1,1,2))

cv.polylines(img4, [pts], False, (0, 0, 225), 7)

cv.line(img4, (180, 210), (320, 210), (0, 0, 225), 7)

plt.imshow(cv.cvtColor(img4, cv.COLOR_BGR2RGB))
plt.show()
```
