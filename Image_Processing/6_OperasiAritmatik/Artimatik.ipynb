```python
import cv2 as cv
import numpy as np
import matplotlib.pyplot as plt
```

```python
image = cv.imread('img.jpg')
```

```python
# Create a matrix of ones, then multiply it by a scaler of 100 
# This gives a matrix with same dimesions of our image with all values being 100

M = np.ones(image.shape, dtype = "uint8")*90
```

```python
# We use this to add this matrix M, to our image
# Notice the increase in brightness

added = cv.add(image, M)
```

```python
# the decrease in brightness
substracted = cv.subtract(image, M)
```

```python
plt.figure(figsize=(20,10))

plt.subplot(1,3,1)
plt.imshow(cv.cvtColor(image, cv.COLOR_BGR2RGB))
plt.axis('off')
plt.title('Normal')

plt.subplot(1,3,2)
plt.imshow(cv.cvtColor(added, cv.COLOR_BGR2RGB))
plt.axis('off')
plt.title('Added')

plt.subplot(1,3,3)
plt.imshow(cv.cvtColor(substracted, cv.COLOR_BGR2RGB))
plt.axis('off')
plt.title('Subracted')

plt.show()
```
