```python
image = cv.imread('img.jpg')
```

```python
plt.figure(figsize=(10,10))

plt.imshow(cv.cvtColor(image, cv.COLOR_BGR2RGB))
plt.axis('off')
plt.show()
```

```python
plt.figure(figsize=(15,10))

#normal image
plt.subplot(2,3,1)
plt.imshow(cv.cvtColor(image, cv.COLOR_BGR2RGB))
plt.title('Original')
plt.axis('off')

#Binary Threshold
plt.subplot(2,3,2)
ret1,thresh1 = cv.threshold(image, 170, 255, cv.THRESH_BINARY)
plt.imshow(cv.cvtColor(thresh1, cv.COLOR_BGR2RGB))
plt.title('Binary')
plt.axis('off')

#Inverted Binary Threshold
plt.subplot(2,3,3)
ret2,thresh2 = cv.threshold(image, 170, 255, cv.THRESH_BINARY_INV)
plt.imshow(cv.cvtColor(thresh2, cv.COLOR_BGR2RGB))
plt.title('Inverted Binary')
plt.axis('off')

#Trunkated Threshold
plt.subplot(2,3,4)
ret3,thresh3 = cv.threshold(image, 170, 255, cv.THRESH_TRUNC)
plt.imshow(cv.cvtColor(thresh3, cv.COLOR_BGR2RGB))
plt.title('Trunkated')
plt.axis('off')

#To zero Threshold
plt.subplot(2,3,5)
ret4, thresh4 = cv.threshold(image, 170,255, cv.THRESH_TOZERO)
plt.imshow(cv.cvtColor(thresh4, cv.COLOR_BGR2RGB))
plt.title('TOZERO')
plt.axis('off')

#Inverted To Zero Threshold
plt.subplot(2,3,6)
ret5, thresh5 = cv.threshold(image, 170, 255, cv.THRESH_TOZERO_INV)
plt.imshow(cv.cvtColor(thresh5, cv.COLOR_BGR2RGB))
plt.title('Inverted TOZERO')
plt.axis('off')

plt.show()
```

## Adaptive Thresholding

```python
# read image as gray
image = cv.imread('img.jpg', 0)
```

```python
plt.figure(figsize=(10,10))

# blur image untuk remove noise
Gauss = cv.GaussianBlur(image, (3,3), 0)

# gunakan adaptiveThreshold
thresh = cv.adaptiveThreshold(Gauss, 255, cv.ADAPTIVE_THRESH_MEAN_C, cv.THRESH_BINARY, 3, 5)

#Normal image
plt.subplot(2,1,1)
plt.imshow(cv.cvtColor(image, cv.COLOR_BGR2RGB))
plt.title('Original')
plt.axis('off')

plt.subplot(2,1,2)
plt.imshow(cv.cvtColor(thresh, cv.COLOR_BGR2RGB))
plt.title('Adaptive Mean Thresholding')
plt.axis('off')

plt.show()
```
