```python
image = cv.imread('anime.jpg')
```

```python
height, width = image.shape[:2]
```

```python
#devide by 2 untuk rotasi gambar di sekitar tengah

rotation_matrix = cv.getRotationMatrix2D((width/2, height/2), 90, .5)
rotated_image = cv.warpAffine(image, rotation_matrix, (width, height))
```

```python
plt.subplot(1,2,1)
plt.imshow(cv.cvtColor(image, cv.COLOR_BGR2RGB))
plt.axis('off')
plt.title('Normal')

plt.subplot(1,2,2)
plt.imshow(cv.cvtColor(rotated_image,cv.COLOR_BGR2RGB))
plt.axis('off')
plt.title('Rotated Image')

plt.tight_layout()
plt.show()
```

```python
rotated_image = cv.transpose(image)

plt.imshow(cv.cvtColor(rotated_image, cv.COLOR_BGR2RGB))
plt.title('Rotated Image')
plt.axis('off')

plt.show()
```

```python
plt.figure(figsize=(12,8))
plt.subplot(1,3,1)
plt.imshow(cv.cvtColor(image, cv.COLOR_BGR2RGB))
plt.axis('off')
plt.title('Normal')

plt.subplot(1,3,2)
hflip_img = cv.flip(image,1)
plt.imshow(cv.cvtColor(hflip_img, cv.COLOR_BGR2RGB))
plt.axis('off')
plt.title('Horizontal Flip')

plt.subplot(1,3,3)
vflip_img = cv.flip(image, 0)
plt.imshow(cv.cvtColor(vflip_img, cv.COLOR_BGR2RGB))
plt.axis('off')
plt.title('Vertical Flip')

plt.show()
```
