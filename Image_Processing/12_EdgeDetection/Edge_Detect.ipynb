```python
import cv2 as cv
import numpy as np
import matplotlib.pyplot as plt
```

# Edge Detection

```python
image = cv.imread('img.jpg')
```

```python
plt.figure(figsize=(10,10))

plt.subplot(2,1,1)
plt.imshow(cv.cvtColor(image, cv.COLOR_BGR2RGB))
plt.title('Original')
plt.axis('off')

#Canny detection menggunakan gradient values as thresholds
plt.subplot(2,1,2)
canny = cv.Canny(image, 100, 200)
plt.imshow(cv.cvtColor(canny, cv.COLOR_BGR2RGB))
plt.title('Canny')
plt.axis('off')

plt.show()
```
