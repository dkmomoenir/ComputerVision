```python
import cv2 as cv
import numpy as np
import matplotlib.pyplot as plt
```

```python
#read image -> gray (binary)
image = cv.imread('img.jpg',0)
```

```python
plt.figure(figsize=(8,8))

plt.imshow(cv.cvtColor(image, cv.COLOR_BGR2RGB))
plt.title('Original Image')
plt.axis('off')

plt.show()
```

# Dilation
- Dilatasi increases the white region in the image

```python
#create kernal
kernel = np.ones((3,3), np.uint8)
```

```python
#define kernel size
kernel = np.ones((5,5), np.uint8)
```

```python
plt.figure(figsize=(8,8))

dilation = cv.dilate(image, kernel, iterations = 1)
plt.imshow(cv.cvtColor(dilation, cv.COLOR_GRAY2RGB))
plt.title('Dilation')
plt.axis('off')

plt.show()
```

# Erosion
- Erosion berguna untuk menghilangkan white noises yang kecil tapi mengecilkan objek

```python
plt.figure(figsize=(8,8))

erosion = cv.erode(image, kernel, iterations = 1)
plt.imshow(cv.cvtColor(erosion, cv.COLOR_GRAY2RGB))
plt.title('Erosion')
plt.axis('off')

plt.show()
```

### Erosion diikuti Dilatasi karena noise juga hilang,noise tak ada kembali, tapi objek area bertambah.

```python
plt.figure(figsize=(8,8))

erosion = cv.erode(image, kernel, iterations = 1)
dilation = cv.dilate(erosion, kernel, iterations = 1)

plt.imshow(cv.cvtColor(dilation, cv.COLOR_GRAY2RGB))
plt.title('Enhanced Image')
plt.axis('off')

plt.show()
```

## Opening
- Good for removing noise

```python
plt.figure(figsize=(8,8))

opening = cv.morphologyEx(image, cv.MORPH_OPEN, kernel)
plt.imshow(cv.cvtColor(opening, cv.COLOR_GRAY2RGB))
plt.title('Opening')
plt.axis('off')

plt.show()
```

## Closing
- Good for removing noise

```python
plt.figure(figsize=(8,8))

closing = cv.morphologyEx(image, cv.MORPH_CLOSE, kernel)

plt.imshow(cv.cvtColor(closing, cv.COLOR_BGR2RGB))
plt.title('Closing')
plt.axis('off')

plt.show()
```
