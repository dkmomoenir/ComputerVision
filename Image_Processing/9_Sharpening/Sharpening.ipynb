```python
import cv2 as cv
import numpy as np
import matplotlib.pyplot as plt
```

Sharpening

By altering our kernels we can implement sharpening, which has the effects of in strengthening or emphasizing edges in an image.

```python
image = cv.imread('img.jpg')
```

```python
#create shapening kernel

kernel_sharpening = np.array([[-1,-1,-1],
                             [-1,9,-1],
                             [-1,-1,-1]])
sharpened = cv.filter2D(image, -1, kernel_sharpening)
```

```python
plt.figure(figsize=15,10))

plt.subplot(1,2,1)
plt.imshow(cv.cvtColor(image, cv.COLOR_BGR2RGB))
plt.axis('off')
plt.title('Normal')

plt.subplot(1,2,2)
plt.imshow(cv.cvtColor(sharpened, cv.COLOR_BGR2RGB))
plt.axis('off')
plt.title
```
