```python
import cv2 as cv
import numpy as np
import matplotlib.pyplot as plt
```

```python
image = cv.imread('anime.jpg')
```

```python
plt.figure(figsize=(10,10))

plt.subplot(2,2,1)
plt.imshow(cv.cvtColor(image, cv.COLOR_BGR2RGB))
plt.axis('off')
plt.title('Normal')

plt.subplot(2,2,2)
image_scaled = cv.resize(image, None, fx=0.15, fy=0.15)
plt.imshow(cv.cvtColor(image_scaled, cv.COLOR_BGR2RGB))
plt.axis('off')
plt.title('Scaling - Linear Interpolation')

plt.subplot(2,2,3)
image_scaled_v1 = cv.resize(image, None, fx=2, fy=2, interpolation = cv.INTER_CUBIC)
plt.imshow(cv.cvtColor(image_scaled_v1, cv.COLOR_BGR2RGB))
plt.axis('off')
plt.title('Scaling - Cubic Interpolation')

plt.subplot(2,2,4)
image_scaled_v2 = cv.resize(image, None, fx=2, fy=2, interpolation = cv.INTER_AREA)
plt.imshow(cv.cvtColor(image_scaled_v2, cv.COLOR_BGR2RGB))
plt.axis('off')
plt.title('Scaling - Area Interpolation')

plt.tight_layout()
plt.show()
```
