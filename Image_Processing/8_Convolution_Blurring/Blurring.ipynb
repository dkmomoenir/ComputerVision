```python
import cv2 as cv
import numpy as np
import matplotlib.pyplot as plt
```

```python
image = cv.imread('image.jpg', 1)
```

```python
plt.figure(figsize=(7,10))

ker = plt.imread('kernals.jpg', 1)
plt.imshow(cv.cvtColor(ker, cv.COLOR_BGR2RGB))
plt.axis('off')
plt.title('Kernels Matrices')
plt.show()
```


Convolutions Bluring (Averaging)
Averaging done by convolving the image with a normalized box filter.
This takes the pixels under the box and replaces the central element.
Box size needs to odd and positive

```python
#box blur kernels

kernel_3x3 = np.ones((3,3), np.float32) / 9
kernel_7x7 = np.ones((7,7), np.float32) / 49
```

```python
plt.figure(figsize=(10,10))

plt.imshow(cv.cvtColor(image, cv.COLOR_BGR2RGB))
plt.axis('off')
plt.title('Normal')
plt.show()
```

```python
#kernel 3x3

plt.figure(figsize=(10,10))

blur3_img = cv.filter2D(image, -1, kernel_3x3)
plt.imshow(cv.cvtColor(blur3_img, cv.COLOR_BGR2RGB))
plt.axis('off')
plt.title('Kernel 3x3')
plt.show()
```

```python
#kernel 7x7

plt.figure(figsize=(10,10))

blur7_img = cv.filter2D(image, -1, kernel_7x7)
plt.imshow(cv.cvtColor(blur7_img, cv.COLOR_BGR2RGB))
plt.axis('off')
plt.title('Kernal 7x7')
plt.show()
```

NOTE: We can use cv2.blur() insted of cv2.filter2D in averaging blur [cv2.blur(image, (3,3))]

Gaussian Blur

```python
#Gaussian 7x7

plt.figure(figsize=(10,10))

Gaussian = cv.GaussianBlur(image, (7,7), 0)
plt.imshow(cv.cvtColor(Gaussian, cv.COLOR_BGR2RGB))
plt.axis('off')
plt.title('Gaussian Blur')

plt.show()
```

Median Blur

```python
# Median 7x7
# Takes median of all the pixels under kernel area and central element is replaced with this median value.

plt.figure(figsize=(10,10))

Median = cv.medianBlur(image, 5)
plt.imshow(cv.cvtColor(Median, cv.COLOR_BGR2RGB))
plt.axis('off')
plt.title('Median Blur')

plt.show()
```

Bilateral Filter

```python
#Gaussian 7x7
#Bilateral is very effective in noise removal while keeping edges sharp

plt.figure(figsize=(10,10))

Bilateral = cv.bilateralFilter(image, 9,75,75)
plt.imshow(cv.cvtColor(Bilateral, cv.COLOR_BGR2RGB))
plt.axis('off')
plt.title('Bilateral Blur')

plt.show()
```

Image De-noising - Non - Local Means Denoising

```python
# Parameters, after None are - the filter strength 'h' (5-10 is a good range)
# Next is hForColorComponents, set as same value as h again

plt.figure(figsize=(10,10))

dst = cv.fastNlMeansDenoisingColored(image, None, 6,6,7,21)
plt.imshow(cv.cvtColor(dst,cv.COLOR_BGR2RGB))
plt.axis('off')
plt.title('Fast Means Denoising')

plt.show()
```

There are 4 variations of Non-Local Means Denoising:

cv2.fastNlMeansDenoising() - works with a single grayscale images
cv2.fastNlMeansDenoisingColored() - works with a color image.
cv2.fastNlMeansDenoisingMulti() - works with image sequence captured in short period of time (grayscale images)
cv2.fastNlMeansDenoisingColoredMulti() - same as above, but for color images.
