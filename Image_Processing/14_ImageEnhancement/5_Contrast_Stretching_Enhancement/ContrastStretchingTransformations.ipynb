```python
import cv2 as cv
import numpy as np
import matplotlib.pyplot as plt
```

```python
#gray
img = cv.imread('image.jpg',0)
rows, columns = img.shape
img_calc = img.astype(np.int32)
```

```python
new_img = np.zeros_like(img)
```

# Contrast Stretching

```python
# (r1, s1)    (r2, s2)    (r3, s3)    (r4, s4)
# (0, 0)      (70, 100)   (120, 200)  (255, 255)
# Constrast_Stretch => if img[r, c] >= r1 and img[r, c] <= r2: r-r1 * s2-s1 / r2-r1 + s1
```

```python
for r in range(rows):
    for c in range(columns):

        pixel = img_calc[r, c]

        # Segment 1: (0,0) -> (70,100)
        if 0 <= pixel <= 70:
            new_img[r, c] = round(
                ((pixel - 0) * (100 - 0)) / (70 - 0) + 0
            )

        # Segment 2: (70,100) -> (120,200)
        elif 70 < pixel <= 120:
            new_img[r, c] = round(
                ((pixel - 70) * (200 - 100)) / (120 - 70) + 100
            )

        # Segment 3: (120,200) -> (255,255)
        elif 120 < pixel <= 255:
            new_img[r, c] = round(
                ((pixel - 120) * (255 - 200)) / (255 - 120) + 200
            )
```

```python
#jangan gunakan cv.cvtColor() karena gambar sudah grayscale
```

```python
plt.figure(figsize=(12, 6))

plt.subplot(1, 2, 1)
plt.imshow(img, cmap='gray')
plt.title('Original Image')
plt.axis('off')

plt.subplot(1, 2, 2)
plt.imshow(new_img, cmap='gray')
plt.title('Contrast Stretched Image')
plt.axis('off')

plt.show()
```
