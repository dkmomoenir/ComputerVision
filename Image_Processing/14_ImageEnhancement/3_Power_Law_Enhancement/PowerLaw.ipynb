```python
import cv2 as cv
import numpy as np
import matplotlib.pyplot as plt
```

```python
image = cv.imread('image.jpg')
```

```python
gamma_list = [0.1, 0.5, 1.2, 2.2]
```

```python
# Convert image to float to prevent uint8 overflow
image_float = image.astype(np.float32)
```

```python
# Apply log transformation
c = 255 / np.log(1 + np.max(image_float))
```

```python
plt.figure(figsize=(12,8))
```

```python
plt.figure(figsize=(16, 4))

for i, gamma in enumerate(gamma_list):
    plt.subplot(1, 4, i + 1)

    gamma_corrected = np.array(
        c * (image / 255) ** gamma,
        dtype='uint8'
    )

    plt.imshow(cv.cvtColor(gamma_corrected, cv.COLOR_BGR2RGB))
    plt.title(
        f'Gamma = {gamma}',
        fontsize=10,
        pad=10
    )
    plt.axis('off')

plt.tight_layout(pad=2)

plt.show()
```
