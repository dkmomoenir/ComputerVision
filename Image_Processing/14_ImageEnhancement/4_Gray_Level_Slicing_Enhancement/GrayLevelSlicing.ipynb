```python
import cv2 as cv
import numpy as np
import matplotlib.pyplot as plt
```

```python
image = cv.imread('image.jpg', 0)
```

```python
rows, columns = image.shape
```

```python
new_img = np.array(image)
```

# Gray Level Slicing

```python
start, end = 100,200
new_val = 255
```

```python
for r in range (rows):
    for c in range(columns):
        if(image[r,c] >= start and image [r,c] <= end):
            new_img[r,c] = new_val
        else:
            new_img[r,c] = 0
```

```python
plt.figure(figsize=(12,8))

plt.subplot(1,2,1)
plt.imshow(cv.cvtColor(image, cv.COLOR_BGR2RGB))
plt.title('Original')
plt.axis('off')

plt.subplot(1,2,2)
plt.imshow(cv.cvtColor(new_img, cv.COLOR_BGR2RGB))
plt.title('Enhanced')
plt.axis('off')

plt.show()
```
