Translasi

```python
img = plt.imread('anime.jpg')
```

```python
height, width = img.shape[:2]
quarter_height, quarter_width = height/4, width/4
```

```python
#matrix translasi

trans = np.float32([[1.2,0, quarter_width-50],
                   [0, 0.9, quarter_height-50]])
```

```python
#gunakan fungsi cv.warpAffine

img_translasi = cv.warpAffine(img, trans, (width, height))
```

```python
plt.figure(figsize=(12,12))
plt.subplot(1,2,1)
plt.imshow(img)
plt.title('Normal')
plt.axis('off')

plt.subplot(1,2,2)
plt.imshow(img_translasi)
plt.axis('off')
plt.title('Translasi')

plt.tight_layout()
plt.show()
```

Shearing

```python
#searing matrix

Shear_mat = np.float32([[1, 0.2, 0],
                        [0.3, 1, 0]])
```

```python
Gambar_shearing = cv.warpAffine(img, Shear_mat, (width, height))
```

```python
plt.figure(figsize=(12,12))
plt.subplot(1,2,1)
plt.imshow(img)
plt.axis('off')
plt.title('Normal')

plt.subplot(1,2,2)
plt.imshow(Gambar_shearing)
plt.title('Shearing')
plt.axis('off')

plt.tight_layout()
plt.show()
```
