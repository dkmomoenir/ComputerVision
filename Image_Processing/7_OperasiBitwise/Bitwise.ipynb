```python
import cv2 as cv
import numpy as np 
import matplotlib.pyplot as plt
```

```python
plt.figure(figsize=(10,10))

plt.subplot(1,2,1)
#persegi
img_rec = np.zeros((512,512), np.uint8)
cv.rectangle(img_rec, (110,50), (400,300), 255, -1)
plt.imshow(cv.cvtColor(img_rec, cv.COLOR_BGR2RGB))
plt.title('Rectangle')
plt.axis('off')

plt.subplot(1,2,2)
#ellipse
img_eli = np.zeros((512,512), np.uint8)
cv.ellipse(img_eli, (170,170), (150,120), 45, 0, 180, 100, -1)
plt.imshow(cv.cvtColor(img_eli, cv.COLOR_BGR2RGB))
plt.title('Rectangle')
plt.axis('off')
      
plt.show()
```

```python
plt.figure(figsize=(10,10))

plt.subplot(1,2,1)
image = np.zeros((512,512), np.uint8)
cv.ellipse(image, (170,170), (150,120), 45, 0, 180, 100, -1)
cv.rectangle(image, (110,50), (400,300), 255, -1)
plt.imshow(cv.cvtColor(image, cv.COLOR_BGR2RGB))
plt.axis('off')

plt.subplot(1,2,2)
image2 = np.zeros((512,512), np.uint8)
cv.rectangle(image2, (110,50), (400,300), 255, -1)
cv.ellipse(image2, (170,170), (150,120), 45, 0, 180, 100, -1)
plt.imshow(cv.cvtColor(image2, cv.COLOR_BGR2RGB))
plt.axis('off')

plt.show()
```

```python
#INTERSECTION

plt.figure(figsize=(8,8))

diantara = cv.bitwise_and(img_rec,img_eli)
plt.imshow(cv.cvtColor(diantara, cv.COLOR_BGR2RGB))
plt.title('Operator AND')
plt.show()
```

```python
#UNION
plt.figure(figsize=(8,8))

atau = cv.bitwise_or(img_rec, img_eli)
plt.imshow(cv.cvtColor(atau, cv.COLOR_BGR2RGB))
plt.title('Operator OR')
plt.show()
```

```python
#XOR

plt.figure(figsize=(8,8))

xor = cv.bitwise_xor(img_rec,img_eli)
plt.imshow(cv.cvtColor(xor, cv.COLOR_BGR2RGB))
plt.title('Operasi XOR')
plt.show()
```

```python
#NOT
plt.figure(figsize=(8,8))

nt = cv.bitwise_not(img_rec)
plt.imshow(cv.cvtColor(nt, cv.COLOR_BGR2RGB))
plt.title('NOT Rectangle')
plt.show()
```
