```python
import cv2 as cv
import numpy as np
import matplotlib.pyplot as plt
```

Pyramid

it deals with resolution of image
use pyrDown() at first, then use pyrUp()

```python
image = cv.imread('wbb.jpg')
```

```python
smaller = cv.pyrDown(image)
larger = cv.pyrUp(image)
```

```python
plt.figure(figsize=(15,15))

plt.subplot(1,3,1)
plt.imshow(cv.cvtColor(image, cv.COLOR_BGR2RGB))
plt.axis('off')
plt.title('Normal')

plt.subplot(1,3,2)
plt.imshow(cv.cvtColor(smaller, cv.COLOR_BGR2RGB))
plt.axis('off')
plt.title('Reduce Resolution')

plt.subplot(1,3,3)
plt.imshow(cv.cvtColor(larger, cv.COLOR_BGR2RGB))
plt.axis('off')
plt.title('Increase Resolution')

plt.tight_layout()
plt.show()
```
